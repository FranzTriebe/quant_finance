# Black-Scholes Model

Prices a European option with the Black-Scholes formula. Running the inputs cell asks for:

- volatility (decimal)
- current share price
- strike price
- interest rate (percent)
- maturity (years)

**Status:** work in progress. Input handling and the `d1`/`d2` calculations are written; the final price formula is not finished yet.
